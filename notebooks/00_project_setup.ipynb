```python
from pathlib import Path

ROOT = Path('/content/drive/MyDrive/HP-SS-Engine')
folders = [
    'notebooks', 'src', 'data/raw', 'data/processed',
    'data/embeddings', 'benchmarks/raw_results',
    'benchmarks/profiling', 'benchmarks/analysis',
    'figures', 'report'
]
for folder in folders:
    (ROOT / folder).mkdir(parents=True, exist_ok=True)
for filename in ['README.md', 'requirements.txt']:
    (ROOT / filename).touch(exist_ok=True)
print('Project folders ready:', ROOT)
```
